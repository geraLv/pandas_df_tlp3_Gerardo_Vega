```python
import pandas as pd
import sqlite3
import numpy as np

#Aqui agrega la conexión a la base de datos
conn = sqlite3.connect("Salaries.sqlite")
df = pd.read_sql_query("SELECT * FROM salaries", conn)
conn.close
df.head(5)
#Crea el dataframe con los datos del datasets
```

1. **Top 10 empleados con mayor salario total (incluyendo beneficios)**
Extrae e imprime un top 10 de expleados con mayor salario incluyendo los beneficios 


```python
#todo
#Filtra los datos por la columna TotalPayBenefits y lo muestra en orden descendente
df_top_salarios = df.sort_values(by="TotalPayBenefits", ascending=False).head(10)
df_top_salarios
```

2. **Filtrar empleados con más de 50,000 en horas extra, la columna a tratar es: (OvertimePay)**

```python
#todo
#Renplaza los valores "Not Provided" por "nan", 
#para poder pasar la columna a float
df["OvertimePay"] = df["OvertimePay"].replace("Not Provided", np.nan)

#Combierte la columna "OvertimePay" a float
df["OvertimePay"] = df["OvertimePay"].astype(float)

#Muestra solo los empleados con más de 50000 horas extra
df_horas_extra = df[df["OvertimePay"] > 50000]
df_horas_extra.head(10)
```

3. **Contar cuántos empleados únicos hay por año**

```python
#todo
#No entendi la consigna
df_empleados_año = df["Year"].value_counts()
df_empleados_año
```

4. **Ver cuántos cargos únicos (JobTitle) existen y los 5 más comunes**

```python
#todo
#Muestra los primeros 5 trabajos más comunes
df_cargos_comunes = df["JobTitle"].value_counts()
print("Trabajos mas comunes:")
print(df_cargos_comunes.head(5))

#Muestra la cantidad de cargos que se repiten una vez
df_cargos_unicos = (df["JobTitle"].value_counts() == 1).value_counts()
print("Hay 239 cargos unicos:")
print(df_cargos_unicos)
```

5. **Mostrar el salario total (TotalPay) promedio por año**

```python
#todo
#Agrupa el salario promedio por año
df_promedio_año = df.groupby("Year")["TotalPay"].mean().reset_index()

#Reemplaza el nombre de las columnas por unos más descriptivos
df_promedio_año.columns = ["Año","      Salario promedio por año"]
print(df_promedio_año)
```
